```python
import shutil
import random

# ============================================================
# EXDARK -> YOLO DATASET CONVERSION
# ============================================================

# -----------------------------
# 1. PATHS
# -----------------------------
IMAGE_DIR = Path(r"D:\ExDark\ExDark")
ANNO_DIR = Path(r"D:\ExDark_Anno\ExDark_Anno")
OUTPUT_DIR = Path(r"D:\ExDark_YOLO")

# -----------------------------
# 2. CLASS DEFINITIONS
# -----------------------------
CLASS_NAMES = [
    "Bicycle",
    "Boat",
    "Bottle",
    "Bus",
    "Car",
    "Cat",
    "Chair",
    "Cup",
    "Dog",
    "Motorbike",
    "People",
    "Table"
]

CLASS_TO_ID = {
    name: idx
    for idx, name in enumerate(CLASS_NAMES)
}

IMAGE_EXTENSIONS = {".jpg", ".jpeg", ".png"}

# Reproducible split
RANDOM_SEED = 42

TRAIN_RATIO = 0.70
VALID_RATIO = 0.15
TEST_RATIO = 0.15

# -----------------------------
# 3. BASIC CHECKS
# -----------------------------
print("=" * 70)
print("EXDARK -> YOLO CONVERSION")
print("=" * 70)

print("\nChecking directories...")

if not IMAGE_DIR.exists():
    raise FileNotFoundError(f"Image directory not found:\n{IMAGE_DIR}")

if not ANNO_DIR.exists():
    raise FileNotFoundError(f"Annotation directory not found:\n{ANNO_DIR}")

print("Images:", IMAGE_DIR)
print("Annotations:", ANNO_DIR)

# -----------------------------
# 4. COLLECT IMAGES
# -----------------------------
print("\nCollecting images...")

images = [
    p for p in IMAGE_DIR.rglob("*")
    if p.is_file() and p.suffix.lower() in IMAGE_EXTENSIONS
]

print("Total images found:", len(images))

# -----------------------------
# 5. BUILD ANNOTATION INDEX
# -----------------------------
print("\nIndexing annotations...")

annotations = {}

for p in ANNO_DIR.rglob("*.txt"):
    # Annotation is named:
    # 2015_00001.jpg.txt
    # 2015_00002.png.txt
    #
    # Remove only the final ".txt"
    image_name = p.name[:-4]

    # Use filename as key
    annotations[image_name] = p

print("Total annotation files found:", len(annotations))
```

```python
# -----------------------------
# 6. MATCH IMAGES + ANNOTATIONS
# -----------------------------
print("\nMatching images with annotations...")

matched = []
missing_annotations = []

for image_path in images:

    image_name = image_path.name

    if image_name in annotations:
        matched.append(
            (image_path, annotations[image_name])
        )
    else:
        missing_annotations.append(image_path)

print("Matched image/annotation pairs:", len(matched))
print("Images without annotations:", len(missing_annotations))

if missing_annotations:
    print("\nFirst missing annotations:")
    for p in missing_annotations[:20]:
        print(" ", p)

# -----------------------------
# 7. PARSE ANNOTATIONS
# -----------------------------
print("\nParsing annotations...")

valid_samples = []
invalid_samples = []

class_distribution = Counter()

for image_path, anno_path in matched:

    try:
        # Read image dimensions without requiring OpenCV
        from PIL import Image

        with Image.open(image_path) as img:
            img_width, img_height = img.size

    except Exception as e:
        invalid_samples.append(
            (image_path, f"Cannot read image: {e}")
        )
        continue

    try:
        lines = anno_path.read_text(
            encoding="utf-8",
            errors="ignore"
        ).splitlines()

    except Exception as e:
        invalid_samples.append(
            (image_path, f"Cannot read annotation: {e}")
        )
        continue

    yolo_lines = []

    for line in lines:

        line = line.strip()

        # Skip empty lines
        if not line:
            continue

        # Skip bbGt header
        if line.startswith("%"):
            continue

        parts = line.split()

        # We expect at least:
        # class left top width height
        if len(parts) < 5:
            invalid_samples.append(
                (image_path, f"Invalid annotation line: {line}")
            )
            continue

        class_name = parts[0]

        if class_name not in CLASS_TO_ID:
            invalid_samples.append(
                (image_path, f"Unknown class: {class_name}")
            )
            continue

        try:
            left = float(parts[1])
            top = float(parts[2])
            width = float(parts[3])
            height = float(parts[4])

        except ValueError:
            invalid_samples.append(
                (image_path, f"Invalid coordinates: {line}")
            )
            continue

        # Validate box
        if width <= 0 or height <= 0:
            invalid_samples.append(
                (image_path, f"Invalid box size: {line}")
            )
            continue

        # Convert:
        #
        # ExDark:
        # left, top, width, height
        #
        # YOLO:
        # class x_center y_center width height

        x_center = left + width / 2
        y_center = top + height / 2

        # Normalize
        x_center /= img_width
        y_center /= img_height
        width_norm = width / img_width
        height_norm = height / img_height

        # Check normalized values
        if not (
            0 <= x_center <= 1
            and 0 <= y_center <= 1
            and 0 < width_norm <= 1
            and 0 < height_norm <= 1
        ):
            invalid_samples.append(
                (
                    image_path,
                    f"Box outside image bounds: {line}"
                )
            )
            continue

        class_id = CLASS_TO_ID[class_name]

        yolo_lines.append(
            f"{class_id} "
            f"{x_center:.6f} "
            f"{y_center:.6f} "
            f"{width_norm:.6f} "
            f"{height_norm:.6f}"
        )

        class_distribution[class_name] += 1

    if yolo_lines:
        valid_samples.append(
            {
                "image": image_path,
                "annotation": anno_path,
                "labels": yolo_lines
            }
        )

print("\nValid samples:", len(valid_samples))
print("Invalid samples/issues:", len(invalid_samples))

# -----------------------------
# 8. CHECK DATASET SIZE
# -----------------------------
if len(valid_samples) == 0:
    raise RuntimeError("No valid samples found.")
```

```python
# -----------------------------
# BUILD CASE-INSENSITIVE ANNOTATION INDEX
# -----------------------------
print("\nIndexing annotations...")

annotations = {}

for p in ANNO_DIR.rglob("*.txt"):

    # Remove final .txt
    image_name = p.name[:-4]

    # Case-insensitive key
    annotations[image_name.lower()] = p

print("Total annotation files found:", len(annotations))


# -----------------------------
# MATCH IMAGES + ANNOTATIONS
# -----------------------------
print("\nMatching images with annotations...")

matched = []
missing_annotations = []

for image_path in images:

    image_name = image_path.name.lower()

    if image_name in annotations:
        matched.append(
            (image_path, annotations[image_name])
        )
    else:
        missing_annotations.append(image_path)

print("Matched image/annotation pairs:", len(matched))
print("Images without annotations:", len(missing_annotations))

if missing_annotations:
    print("\nFirst missing annotations:")
    for p in missing_annotations[:20]:
        print(" ", p)
```

```python
# 9. STRATIFIED SPLIT
# -----------------------------
print("\nCreating stratified train/valid/test split...")

# Group samples by their primary image folder.
# This keeps the original ExDark class distribution represented.
groups = {}

for sample in valid_samples:

    # Folder containing the image
    class_folder = sample["image"].parent.name

    groups.setdefault(class_folder, []).append(sample)

random.seed(RANDOM_SEED)

train_samples = []
valid_samples_split = []
test_samples = []

for class_name, samples in groups.items():

    random.shuffle(samples)

    n = len(samples)

    n_train = int(n * TRAIN_RATIO)
    n_valid = int(n * VALID_RATIO)

    train_part = samples[:n_train]
    valid_part = samples[n_train:n_train + n_valid]
    test_part = samples[n_train + n_valid:]

    train_samples.extend(train_part)
    valid_samples_split.extend(valid_part)
    test_samples.extend(test_part)

print("Train:", len(train_samples))
print("Valid:", len(valid_samples_split))
print("Test :", len(test_samples))
print("Total:", len(train_samples) +
      len(valid_samples_split) +
      len(test_samples))

# -----------------------------
# 10. CREATE OUTPUT DIRECTORIES
# -----------------------------
print("\nCreating YOLO directory structure...")

for split in ["train", "valid", "test"]:

    (OUTPUT_DIR / split / "images").mkdir(
        parents=True,
        exist_ok=True
    )

    (OUTPUT_DIR / split / "labels").mkdir(
        parents=True,
        exist_ok=True
    )

# -----------------------------
# 11. COPY IMAGES + WRITE LABELS
# -----------------------------
def process_split(samples, split_name):

    image_output = OUTPUT_DIR / split_name / "images"
    label_output = OUTPUT_DIR / split_name / "labels"

    for sample in samples:

        source_image = sample["image"]

        destination_image = image_output / source_image.name

        destination_label = (
            label_output /
            f"{source_image.name}.txt"
        )

        shutil.copy2(
            source_image,
            destination_image
        )

        destination_label.write_text(
            "\n".join(sample["labels"]) + "\n",
            encoding="utf-8"
        )

    print(
        f"{split_name}: "
        f"{len(samples)} images + labels created"
    )


process_split(train_samples, "train")
process_split(valid_samples_split, "valid")
process_split(test_samples, "test")

# -----------------------------
# 12. CREATE data.yaml
# -----------------------------
yaml_lines = [
    f"path: {OUTPUT_DIR.as_posix()}",
    "train: train/images",
    "val: valid/images",
    "test: test/images",
    "",
    f"nc: {len(CLASS_NAMES)}",
    "names:"
]

for idx, name in enumerate(CLASS_NAMES):
    yaml_lines.append(f"  {idx}: {name}")

yaml_content = "\n".join(yaml_lines) + "\n"

yaml_path = OUTPUT_DIR / "data.yaml"
yaml_path.write_text(
    yaml_content,
    encoding="utf-8"
)

# -----------------------------
# 13. FINAL SUMMARY
# -----------------------------
print("\n" + "=" * 70)
print("CONVERSION COMPLETE")
print("=" * 70)

print(f"\nOutput directory:")
print(OUTPUT_DIR)

print("\nDataset:")
print(f"  Train: {len(train_samples)}")
print(f"  Valid: {len(valid_samples_split)}")
print(f"  Test : {len(test_samples)}")
print(f"  Total: {len(train_samples) + len(valid_samples_split) + len(test_samples)}")

print("\nClasses:")
for idx, name in enumerate(CLASS_NAMES):
    print(f"  {idx}: {name}")

print("\nObject distribution:")
for name in CLASS_NAMES:
    print(f"  {name:12s}: {class_distribution[name]}")

print("\ndata.yaml:")
print(yaml_path)

print("\n" + "=" * 70)
```

```python

import random
import re

# ============================================================
# PATHS
# ============================================================

IMAGE_DIR = Path(r"D:\ExDark\ExDark")
ANNO_DIR = Path(r"D:\ExDark_Anno\ExDark_Anno")
OUTPUT_DIR = Path(r"D:\ExDark_YOLO")

SEED = 42

# ============================================================
# CLASS DEFINITIONS
# ============================================================

CLASS_NAMES = [
    "Bicycle",
    "Boat",
    "Bottle",
    "Bus",
    "Car",
    "Cat",
    "Chair",
    "Cup",
    "Dog",
    "Motorbike",
    "People",
    "Table"
]

CLASS_TO_ID = {
    name.lower(): i
    for i, name in enumerate(CLASS_NAMES)
}

VALID_IMAGE_EXTENSIONS = {
    ".jpg", ".jpeg", ".png",
    ".JPG", ".JPEG", ".PNG"
}

# ============================================================
# CLEAN PREVIOUS OUTPUT
# ============================================================

if OUTPUT_DIR.exists():
    print(f"Removing previous dataset: {OUTPUT_DIR}")
    shutil.rmtree(OUTPUT_DIR)

# ============================================================
# CREATE YOLO DIRECTORY STRUCTURE
# ============================================================

for split in ["train", "valid", "test"]:
    (OUTPUT_DIR / split / "images").mkdir(parents=True, exist_ok=True)
    (OUTPUT_DIR / split / "labels").mkdir(parents=True, exist_ok=True)

print("YOLO directory structure created.")

# ============================================================
# COLLECT ORIGINAL IMAGES
# ============================================================

print("\nCollecting images...")

images = [
    p for p in IMAGE_DIR.rglob("*")
    if p.is_file() and p.suffix in VALID_IMAGE_EXTENSIONS
]

print(f"Total original images found: {len(images)}")

# ============================================================
# COLLECT ANNOTATIONS
# ============================================================

print("\nCollecting annotations...")

annotations = [
    p for p in ANNO_DIR.rglob("*.txt")
    if p.is_file()
]

print(f"Total annotation files found: {len(annotations)}")

# ============================================================
# BUILD CASE-INSENSITIVE ANNOTATION INDEX
# ============================================================

print("\nBuilding annotation index...")

annotation_index = {}

for anno in annotations:

    # Example:
    # 2015_00391.jpg.txt
    filename = anno.name.lower()

    # Remove .txt
    if filename.endswith(".txt"):
        filename = filename[:-4]

    # Remove image extension if present
    filename = Path(filename).stem.lower()

    annotation_index[filename] = anno

print(f"Indexed annotations: {len(annotation_index)}")

# ============================================================
# MATCH IMAGES TO ANNOTATIONS
# ============================================================

print("\nMatching images with annotations...")

matched_pairs = []
missing_annotations = []

for image in images:

    image_stem = image.stem.lower()

    anno = annotation_index.get(image_stem)

    if anno is None:
        missing_annotations.append(image)
    else:
        matched_pairs.append((image, anno))

print(f"Matched image/annotation pairs: {len(matched_pairs)}")
print(f"Images without annotations: {len(missing_annotations)}")

if missing_annotations:
    print("\nMissing annotations:")
    for image in missing_annotations:
        print(image)

# ============================================================
# PARSE ANNOTATIONS
# ============================================================

print("\nParsing annotations...")

valid_pairs = []
invalid_pairs = []

for image, anno in matched_pairs:

    try:
        with open(anno, "r", encoding="utf-8", errors="ignore") as f:
            lines = f.readlines()

        yolo_lines = []

        for line in lines:

            line = line.strip()

            if not line:
                continue

            # Remove optional ExDark header
            if line.startswith("%"):
                continue

            parts = line.split()

            if len(parts) < 5:
                continue

            class_name = parts[0].lower()

            if class_name not in CLASS_TO_ID:
                raise ValueError(
                    f"Unknown class '{parts[0]}' in {anno}"
                )

            # ExDark format:
            #
            # class left top width height ...
            #
            left = float(parts[1])
            top = float(parts[2])
            width = float(parts[3])
            height = float(parts[4])

            if width <= 0 or height <= 0:
                raise ValueError(
                    f"Invalid bounding box in {anno}"
                )

            class_id = CLASS_TO_ID[class_name]

            # ------------------------------------------------
            # Read image dimensions
            # ------------------------------------------------

            from PIL import Image

            with Image.open(image) as img:
                img_width, img_height = img.size

            # ------------------------------------------------
            # Convert:
            #
            # left, top, width, height
            #
            # to YOLO:
            #
            # x_center, y_center, width, height
            #
            # normalized 0-1
            # ------------------------------------------------

            x_center = (left + width / 2) / img_width
            y_center = (top + height / 2) / img_height

            norm_width = width / img_width
            norm_height = height / img_height

            # Validate normalized coordinates
            if not (
                0 <= x_center <= 1
                and 0 <= y_center <= 1
                and 0 < norm_width <= 1
                and 0 < norm_height <= 1
            ):
                raise ValueError(
                    f"Bounding box outside valid range in {anno}"
                )

            yolo_lines.append(
                f"{class_id} "
                f"{x_center:.6f} "
                f"{y_center:.6f} "
                f"{norm_width:.6f} "
                f"{norm_height:.6f}"
            )

        if not yolo_lines:
            raise ValueError(
                f"No valid annotations found in {anno}"
            )

        valid_pairs.append((image, anno, yolo_lines))

    except Exception as e:

        invalid_pairs.append((image, anno, str(e)))

print(f"Valid samples: {len(valid_pairs)}")
print(f"Invalid samples/issues: {len(invalid_pairs)}")

if invalid_pairs:
    print("\nInvalid samples:")
    for image, anno, error in invalid_pairs[:20]:
        print(f"{image}")
        print(f"  Annotation: {anno}")
        print(f"  Error: {error}")

# ============================================================
# STRATIFIED SPLIT
# ============================================================

print("\nCreating stratified train/valid/test split...")

# Group by original ExDark class folder
groups = {}

for item in valid_pairs:

    image = item[0]

    # Original class folder
    group = image.parent.name

    groups.setdefault(group, []).append(item)

random.seed(SEED)

train_items = []
valid_items = []
test_items = []

for group, items in groups.items():

    random.shuffle(items)

    n = len(items)

    n_train = round(n * 0.70)
    n_valid = round(n * 0.15)

    train_items.extend(items[:n_train])
    valid_items.extend(items[n_train:n_train + n_valid])
    test_items.extend(items[n_train + n_valid:])

# Shuffle final splits
random.shuffle(train_items)
random.shuffle(valid_items)
random.shuffle(test_items)

print(f"Train: {len(train_items)}")
print(f"Valid: {len(valid_items)}")
print(f"Test : {len(test_items)}")
print(f"Total: {len(train_items) + len(valid_items) + len(test_items)}")

# ============================================================
# COPY IMAGES + CREATE YOLO LABELS
# ============================================================

def create_split(items, split_name):

    image_output = OUTPUT_DIR / split_name / "images"
    label_output = OUTPUT_DIR / split_name / "labels"

    for image, anno, yolo_lines in items:

        # Copy original image unchanged
        destination_image = image_output / image.name
        shutil.copy2(image, destination_image)

        # IMPORTANT:
        # Use image.stem so:
        #
        # 2015_00391.JPG
        #
        # becomes:
        #
        # 2015_00391.txt
        #
        label_filename = image.stem + ".txt"

        destination_label = label_output / label_filename

        with open(
            destination_label,
            "w",
            encoding="utf-8"
        ) as f:
            f.write("\n".join(yolo_lines))

    print(
        f"{split_name}: "
        f"{len(items)} images + "
        f"{len(items)} labels created"
    )

# ============================================================
# CREATE DATASET
# ============================================================

create_split(train_items, "train")
create_split(valid_items, "valid")
create_split(test_items, "test")

# ============================================================
# CREATE data.yaml
# ============================================================

yaml_content = f"""path: {OUTPUT_DIR.as_posix()}
train: train/images
val: valid/images
test: test/images

nc: {len(CLASS_NAMES)}

names:
"""

for i, name in enumerate(CLASS_NAMES):
    yaml_content += f"  {i}: {name}\n"

yaml_path = OUTPUT_DIR / "data.yaml"

with open(yaml_path, "w", encoding="utf-8") as f:
    f.write(yaml_content)

# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("CONVERSION COMPLETE")
print("=" * 70)

print(f"Output directory: {OUTPUT_DIR}")
print()
print(f"Train: {len(train_items)}")
print(f"Valid: {len(valid_items)}")
print(f"Test : {len(test_items)}")
print(f"Total: {len(valid_pairs)}")

print()
print(f"data.yaml: {yaml_path}")

print("\nClass mapping:")
for i, name in enumerate(CLASS_NAMES):
    print(f"{i}: {name}")
```

```python

from PIL import Image

# ============================================================
# CONFIGURATION
# ============================================================

IMAGE_DIR = Path(r"D:\ExDark\ExDark")
ANNO_DIR = Path(r"D:\ExDark_Anno\ExDark_Anno")
OUTPUT_DIR = Path(r"D:\ExDark_YOLO")

SEED = 42

# ============================================================
# EXDARK CLASSES
# ============================================================

CLASS_NAMES = [
    "Bicycle",
    "Boat",
    "Bottle",
    "Bus",
    "Car",
    "Cat",
    "Chair",
    "Cup",
    "Dog",
    "Motorbike",
    "People",
    "Table"
]

CLASS_TO_ID = {
    name.lower(): i
    for i, name in enumerate(CLASS_NAMES)
}

IMAGE_EXTENSIONS = {
    ".jpg",
    ".jpeg",
    ".png"
}

# ============================================================
# REMOVE PREVIOUS OUTPUT
# ============================================================

if OUTPUT_DIR.exists():
    print(f"Removing previous dataset:")
    print(OUTPUT_DIR)
    shutil.rmtree(OUTPUT_DIR)

# ============================================================
# CREATE YOLO DIRECTORY STRUCTURE
# ============================================================

for split in ["train", "valid", "test"]:
    (OUTPUT_DIR / split / "images").mkdir(
        parents=True,
        exist_ok=True
    )

    (OUTPUT_DIR / split / "labels").mkdir(
        parents=True,
        exist_ok=True
    )

print("\nYOLO directory structure created.")

# ============================================================
# COLLECT ORIGINAL IMAGES
# ============================================================

print("\n" + "=" * 70)
print("COLLECTING IMAGES")
print("=" * 70)

images = [
    p
    for p in IMAGE_DIR.rglob("*")
    if p.is_file()
    and p.suffix.lower() in IMAGE_EXTENSIONS
]

print(f"Original images found: {len(images)}")

# ============================================================
# COLLECT ANNOTATIONS
# ============================================================

print("\n" + "=" * 70)
print("COLLECTING ANNOTATIONS")
print("=" * 70)

annotations = [
    p
    for p in ANNO_DIR.rglob("*.txt")
    if p.is_file()
]

print(f"Annotation files found: {len(annotations)}")

# ============================================================
# BUILD ANNOTATION INDEX
#
# Converts:
# 2015_00391.jpg.txt
#
# into:
# 2015_00391
#
# This allows matching regardless of JPG/jpg case.
# ============================================================

print("\nBuilding annotation index...")

annotation_index = {}

for anno in annotations:

    filename = anno.name.lower()

    # Remove final .txt
    if filename.endswith(".txt"):
        filename = filename[:-4]

    # Remove image extension
    filename = Path(filename).stem.lower()

    annotation_index[filename] = anno

print(f"Indexed annotations: {len(annotation_index)}")

# ============================================================
# MATCH IMAGES WITH ANNOTATIONS
# ============================================================

print("\n" + "=" * 70)
print("MATCHING IMAGES WITH ANNOTATIONS")
print("=" * 70)

matched_pairs = []
missing_annotations = []

for image in images:

    image_stem = image.stem.lower()

    anno = annotation_index.get(image_stem)

    if anno is None:
        missing_annotations.append(image)
    else:
        matched_pairs.append((image, anno))

print(f"Matched image/annotation pairs: {len(matched_pairs)}")
print(f"Images without annotations: {len(missing_annotations)}")

if missing_annotations:

    print("\nImages without annotation files:")

    for image in missing_annotations:
        print(image)

# ============================================================
# PARSE ANNOTATIONS
# ============================================================

print("\n" + "=" * 70)
print("PARSING ANNOTATIONS")
print("=" * 70)

valid_pairs = []
invalid_pairs = []

for image, anno in matched_pairs:

    try:

        # ----------------------------------------------------
        # Read annotation
        # ----------------------------------------------------

        with open(
            anno,
            "r",
            encoding="utf-8",
            errors="ignore"
        ) as f:
            lines = f.readlines()

        # ----------------------------------------------------
        # Get image dimensions
        # ----------------------------------------------------

        with Image.open(image) as img:
            image_width, image_height = img.size

        yolo_lines = []

        # ----------------------------------------------------
        # Parse every annotation line
        # ----------------------------------------------------

        for line in lines:

            line = line.strip()

            # Empty line
            if not line:
                continue

            # ExDark header
            if line.startswith("%"):
                continue

            parts = line.split()

            # Need at least:
            # class left top width height
            if len(parts) < 5:
                continue

            class_name = parts[0].lower()

            if class_name not in CLASS_TO_ID:
                raise ValueError(
                    f"Unknown class '{parts[0]}'"
                )

            try:
                left = float(parts[1])
                top = float(parts[2])
                width = float(parts[3])
                height = float(parts[4])
            except ValueError:
                raise ValueError(
                    f"Invalid numeric values: {parts}"
                )

            # ------------------------------------------------
            # Validate original bounding box
            # ------------------------------------------------

            if width <= 0 or height <= 0:
                raise ValueError(
                    "Bounding box has non-positive width/height"
                )

            if left < 0 or top < 0:
                raise ValueError(
                    "Bounding box has negative coordinates"
                )

            # ------------------------------------------------
            # Convert ExDark format:
            #
            # left top width height
            #
            # to YOLO:
            #
            # x_center y_center width height
            # ------------------------------------------------

            x_center = left + width / 2
            y_center = top + height / 2

            # ------------------------------------------------
            # Normalize to 0-1
            # ------------------------------------------------

            x_center /= image_width
            y_center /= image_height

            norm_width = width / image_width
            norm_height = height / image_height

            # ------------------------------------------------
            # Validate YOLO coordinates
            # ------------------------------------------------

            if not (
                0 <= x_center <= 1
                and 0 <= y_center <= 1
                and 0 < norm_width <= 1
                and 0 < norm_height <= 1
            ):
                raise ValueError(
                    "Bounding box falls outside valid YOLO range"
                )

            class_id = CLASS_TO_ID[class_name]

            yolo_lines.append(
                f"{class_id} "
                f"{x_center:.6f} "
                f"{y_center:.6f} "
                f"{norm_width:.6f} "
                f"{norm_height:.6f}"
            )

        # ----------------------------------------------------
        # No actual bounding boxes
        #
        # This catches:
        #
        # % bbGt version=3
        #
        # with no object annotations.
        # ----------------------------------------------------

        if len(yolo_lines) == 0:
            raise ValueError(
                f"No valid annotations found in {anno}"
            )

        valid_pairs.append(
            (image, anno, yolo_lines)
        )

    except Exception as e:

        invalid_pairs.append(
            (image, anno, str(e))
        )

print(f"Valid samples: {len(valid_pairs)}")
print(f"Invalid samples/issues: {len(invalid_pairs)}")

if invalid_pairs:

    print("\nInvalid samples:")

    for image, anno, error in invalid_pairs:

        print(f"\nImage: {image}")
        print(f"Annotation: {anno}")
        print(f"Error: {error}")

# ============================================================
# STRATIFIED TRAIN / VALID / TEST SPLIT
#
# Split based on the original ExDark image folders.
#
# 70% train
# 15% valid
# 15% test
# ============================================================

print("\n" + "=" * 70)
print("CREATING STRATIFIED TRAIN/VALID/TEST SPLIT")
print("=" * 70)

groups = {}

for item in valid_pairs:

    image = item[0]

    # Original ExDark class folder
    group = image.parent.name

    groups.setdefault(group, [])

    groups[group].append(item)

random.seed(SEED)

train_items = []
valid_items = []
test_items = []

for group, items in groups.items():

    random.shuffle(items)

    n = len(items)

    n_train = round(n * 0.70)
    n_valid = round(n * 0.15)

    train_items.extend(
        items[:n_train]
    )

    valid_items.extend(
        items[
            n_train:n_train + n_valid
        ]
    )

    test_items.extend(
        items[
            n_train + n_valid:
        ]
    )

# Shuffle final datasets
random.shuffle(train_items)
random.shuffle(valid_items)
random.shuffle(test_items)

print(f"Train: {len(train_items)}")
print(f"Valid: {len(valid_items)}")
print(f"Test : {len(test_items)}")
print(
    f"Total: "
    f"{len(train_items) + len(valid_items) + len(test_items)}"
)

# ============================================================
# COPY IMAGES AND CREATE YOLO LABELS
# ============================================================

print("\n" + "=" * 70)
print("CREATING YOLO DATASET")
print("=" * 70)

def create_split(items, split_name):

    image_dir = (
        OUTPUT_DIR
        / split_name
        / "images"
    )

    label_dir = (
        OUTPUT_DIR
        / split_name
        / "labels"
    )

    for image, anno, yolo_lines in items:

        # ----------------------------------------------------
        # Copy original image unchanged
        # ----------------------------------------------------

        destination_image = (
            image_dir / image.name
        )

        shutil.copy2(
            image,
            destination_image
        )

        # ----------------------------------------------------
        # Correct YOLO label filename
        #
        # 2015_00391.JPG
        #       ↓
        # 2015_00391.txt
        # ----------------------------------------------------

        label_filename = (
            image.stem + ".txt"
        )

        destination_label = (
            label_dir / label_filename
        )

        with open(
            destination_label,
            "w",
            encoding="utf-8"
        ) as f:

            f.write(
                "\n".join(yolo_lines)
            )

    print(
        f"{split_name}: "
        f"{len(items)} images + "
        f"{len(items)} labels created"
    )

# Create all splits
create_split(train_items, "train")
create_split(valid_items, "valid")
create_split(test_items, "test")

# ============================================================
# CREATE data.yaml
# ============================================================

print("\n" + "=" * 70)
print("CREATING data.yaml")
print("=" * 70)

yaml_content = f"""path: {OUTPUT_DIR.as_posix()}
train: train/images
val: valid/images
test: test/images

nc: {len(CLASS_NAMES)}

names:
"""

for i, name in enumerate(CLASS_NAMES):

    yaml_content += (
        f"  {i}: {name}\n"
    )

yaml_path = OUTPUT_DIR / "data.yaml"

with open(
    yaml_path,
    "w",
    encoding="utf-8"
) as f:

    f.write(yaml_content)

print(f"Created: {yaml_path}")

# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("CONVERSION COMPLETE")
print("=" * 70)

print(f"Output directory: {OUTPUT_DIR}")

print()
print(f"Train: {len(train_items)}")
print(f"Valid: {len(valid_items)}")
print(f"Test : {len(test_items)}")
print(
    f"Total: "
    f"{len(train_items) + len(valid_items) + len(test_items)}"
)

print()
print(f"data.yaml: {yaml_path}")

print("\nExcluded samples:")

for image, anno, error in invalid_pairs:
    print(f"  {image}")
    print(f"  Reason: {error}")

print("\nClass mapping:")

for i, name in enumerate(CLASS_NAMES):
    print(f"  {i}: {name}")

print("\n" + "=" * 70)
print("DONE")
print("=" * 70)
```

```python

# ============================================================
# 1. PATHS
# ============================================================

IMAGE_ROOT = Path(r"D:\ExDark\ExDark")
ANNOTATION_ROOT = Path(r"D:\ExDark_Anno\ExDark_Anno")
OUTPUT_ROOT = Path(r"D:\ExDark_YOLO")

TRAIN_IMG = OUTPUT_ROOT / "train" / "images"
TRAIN_LBL = OUTPUT_ROOT / "train" / "labels"

VALID_IMG = OUTPUT_ROOT / "valid" / "images"
VALID_LBL = OUTPUT_ROOT / "valid" / "labels"

TEST_IMG = OUTPUT_ROOT / "test" / "images"
TEST_LBL = OUTPUT_ROOT / "test" / "labels"


# ============================================================
# 2. CLASS DEFINITIONS
# ============================================================

CLASS_NAMES = [
    "Bicycle",
    "Boat",
    "Bottle",
    "Bus",
    "Car",
    "Cat",
    "Chair",
    "Cup",
    "Dog",
    "Motorbike",
    "People",
    "Table"
]

CLASS_TO_ID = {
    name.lower(): i
    for i, name in enumerate(CLASS_NAMES)
}


# ============================================================
# 3. CREATE OUTPUT DIRECTORIES
# ============================================================

for folder in [
    TRAIN_IMG, TRAIN_LBL,
    VALID_IMG, VALID_LBL,
    TEST_IMG, TEST_LBL
]:
    folder.mkdir(parents=True, exist_ok=True)


# ============================================================
# 4. FIND ALL IMAGES
# ============================================================

image_extensions = {
    ".jpg", ".jpeg", ".png", ".bmp", ".webp"
}

images = []

for path in IMAGE_ROOT.rglob("*"):

    # Ignore __MACOSX
    if "__MACOSX" in path.parts:
        continue

    if path.is_file() and path.suffix.lower() in image_extensions:
        images.append(path)


print("=" * 80)
print("EXDARK DATASET CONVERSION")
print("=" * 80)

print(f"Image root       : {IMAGE_ROOT}")
print(f"Annotation root  : {ANNOTATION_ROOT}")
print(f"Output root      : {OUTPUT_ROOT}")
print(f"Original images  : {len(images)}")


# ============================================================
# 5. INDEX ANNOTATIONS
# ============================================================

annotation_index = {}

for path in ANNOTATION_ROOT.rglob("*.txt"):

    if "__MACOSX" in path.parts:
        continue

    # Case-insensitive filename matching
    key = path.name.lower()

    annotation_index[key] = path


print(f"Annotation files : {len(annotation_index)}")


# ============================================================
# 6. PARSE ANNOTATION
# ============================================================

def parse_annotation(annotation_path, img_width, img_height):

    yolo_annotations = []

    with open(
        annotation_path,
        "r",
        encoding="utf-8",
        errors="ignore"
    ) as f:

        lines = f.readlines()

    for line in lines:

        line = line.strip()

        # Skip empty lines
        if not line:
            continue

        # Skip ExDark header
        if line.startswith("%"):
            continue

        parts = line.split()

        # Expected format:
        #
        # class_name left top width height ...
        #
        if len(parts) < 5:
            continue

        class_name = parts[0].strip()

        class_id = CLASS_TO_ID.get(class_name.lower())

        if class_id is None:
            continue

        try:
            left = float(parts[1])
            top = float(parts[2])
            width = float(parts[3])
            height = float(parts[4])

        except ValueError:
            continue

        # ----------------------------------------------------
        # Convert from:
        #
        # x, y, width, height
        #
        # to:
        #
        # x1, y1, x2, y2
        # ----------------------------------------------------

        x1 = left
        y1 = top
        x2 = left + width
        y2 = top + height

        # ----------------------------------------------------
        # CLIP TO IMAGE BOUNDARIES
        # ----------------------------------------------------

        x1 = max(0.0, min(x1, img_width))
        y1 = max(0.0, min(y1, img_height))
        x2 = max(0.0, min(x2, img_width))
        y2 = max(0.0, min(y2, img_height))

        # ----------------------------------------------------
        # Check whether the clipped box is still valid
        # ----------------------------------------------------

        if x2 <= x1 or y2 <= y1:
            continue

        # ----------------------------------------------------
        # Calculate clipped width and height
        # ----------------------------------------------------

        clipped_width = x2 - x1
        clipped_height = y2 - y1

        # ----------------------------------------------------
        # YOLO center coordinates
        # ----------------------------------------------------

        x_center = (x1 + x2) / 2.0
        y_center = (y1 + y2) / 2.0

        # ----------------------------------------------------
        # Normalize to 0-1
        # ----------------------------------------------------

        x_center /= img_width
        y_center /= img_height

        clipped_width /= img_width
        clipped_height /= img_height

        # ----------------------------------------------------
        # Final safety check
        # ----------------------------------------------------

        if not (
            0 <= x_center <= 1
            and 0 <= y_center <= 1
            and 0 < clipped_width <= 1
            and 0 < clipped_height <= 1
        ):
            continue

        # ----------------------------------------------------
        # YOLO format
        # ----------------------------------------------------

        yolo_line = (
            f"{class_id} "
            f"{x_center:.6f} "
            f"{y_center:.6f} "
            f"{clipped_width:.6f} "
            f"{clipped_height:.6f}"
        )

        yolo_annotations.append(yolo_line)

    return yolo_annotations


# ============================================================
# 7. MATCH IMAGES WITH ANNOTATIONS + CONVERT
# ============================================================

valid_pairs = []
invalid_pairs = []

print("\n" + "=" * 80)
print("MATCHING AND CONVERTING")
print("=" * 80)

for i, image_path in enumerate(images, 1):

    # Annotation filenames are like:
    #
    # 2015_04089.jpg.txt
    #
    annotation_name = image_path.name.lower() + ".txt"

    annotation_path = annotation_index.get(annotation_name)

    if annotation_path is None:

        invalid_pairs.append(
            (
                image_path,
                None,
                "Annotation not found"
            )
        )

        continue

    # --------------------------------------------------------
    # Read image dimensions
    # --------------------------------------------------------

    try:

        with Image.open(image_path) as img:
            img_width, img_height = img.size

    except Exception as e:

        invalid_pairs.append(
            (
                image_path,
                annotation_path,
                f"Could not read image: {e}"
            )
        )

        continue

    # --------------------------------------------------------
    # Parse annotation
    # --------------------------------------------------------

    yolo_annotations = parse_annotation(
        annotation_path,
        img_width,
        img_height
    )

    # --------------------------------------------------------
    # No valid objects
    # --------------------------------------------------------

    if len(yolo_annotations) == 0:

        invalid_pairs.append(
            (
                image_path,
                annotation_path,
                "No valid annotations found"
            )
        )

        continue

    # --------------------------------------------------------
    # Keep valid pair
    # --------------------------------------------------------

    valid_pairs.append(
        (
            image_path,
            annotation_path,
            yolo_annotations
        )
    )

    if i % 500 == 0:
        print(f"Processed {i}/{len(images)} images...")


# ============================================================
# 8. SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("CONVERSION SUMMARY")
print("=" * 80)

print(f"Original images      : {len(images)}")
print(f"Valid image pairs    : {len(valid_pairs)}")
print(f"Invalid/excluded     : {len(invalid_pairs)}")


# ============================================================
# 9. SHOW INVALID FILES
# ============================================================

if invalid_pairs:

    print("\n" + "=" * 80)
    print("EXCLUDED SAMPLES")
    print("=" * 80)

    for i, (image, annotation, error) in enumerate(
        invalid_pairs, 1
    ):

        print(f"\n{i}. IMAGE")
        print(image)

        print("   ANNOTATION")
        print(annotation)

        print("   REASON")
        print(error)

else:

    print("\nNo invalid samples.")


# ============================================================
# 10. STRATIFIED SPLIT BY ORIGINAL EXDARK CLASS FOLDER
# ============================================================

random.seed(42)

groups = defaultdict(list)

for item in valid_pairs:

    image_path = item[0]

    # Original ExDark class folder
    class_folder = image_path.parent.name

    groups[class_folder].append(item)


train_items = []
valid_items = []
test_items = []

for class_name, items in groups.items():

    random.shuffle(items)

    n = len(items)

    train_end = int(n * 0.70)
    valid_end = train_end + int(n * 0.15)

    train_items.extend(items[:train_end])
    valid_items.extend(items[train_end:valid_end])
    test_items.extend(items[valid_end:])


# Shuffle final lists
random.shuffle(train_items)
random.shuffle(valid_items)
random.shuffle(test_items)


print("\n" + "=" * 80)
print("DATASET SPLIT")
print("=" * 80)

print(f"Train : {len(train_items)}")
print(f"Valid : {len(valid_items)}")
print(f"Test  : {len(test_items)}")
print(f"Total : {len(train_items) + len(valid_items) + len(test_items)}")


# ============================================================
# 11. COPY IMAGES + WRITE YOLO LABELS
# ============================================================

def write_split(items, image_dir, label_dir):

    for image_path, annotation_path, yolo_annotations in items:

        # ----------------------------------------------------
        # Use image stem
        #
        # 2015_04089.jpg
        #
        # becomes:
        #
        # 2015_04089.jpg
        # 2015_04089.txt
        # ----------------------------------------------------

        image_destination = image_dir / image_path.name
        label_destination = label_dir / f"{image_path.stem}.txt"

        # Copy original image unchanged
        shutil.copy2(
            image_path,
            image_destination
        )

        # Write YOLO annotation
        with open(
            label_destination,
            "w",
            encoding="utf-8"
        ) as f:

            f.write(
                "\n".join(yolo_annotations)
            )


print("\n" + "=" * 80)
print("WRITING DATASET")
print("=" * 80)

write_split(
    train_items,
    TRAIN_IMG,
    TRAIN_LBL
)

write_split(
    valid_items,
    VALID_IMG,
    VALID_LBL
)

write_split(
    test_items,
    TEST_IMG,
    TEST_LBL
)

print("Dataset files written successfully.")


# ============================================================
# 12. CREATE data.yaml
# ============================================================

yaml_path = OUTPUT_ROOT / "data.yaml"

yaml_text = f"""path: {OUTPUT_ROOT.as_posix()}
train: train/images
val: valid/images
test: test/images

nc: {len(CLASS_NAMES)}

names:
"""

for i, name in enumerate(CLASS_NAMES):
    yaml_text += f"  {i}: {name}\n"


with open(
    yaml_path,
    "w",
    encoding="utf-8"
) as f:

    f.write(yaml_text)


print("\n" + "=" * 80)
print("data.yaml CREATED")
print("=" * 80)

print(yaml_path)

print("\n" + yaml_text)


# ============================================================
# 13. FINAL TOTALS
# ============================================================

final_total = (
    len(train_items)
    + len(valid_items)
    + len(test_items)
)

print("=" * 80)
print("FINAL RESULT")
print("=" * 80)

print(f"Original images : {len(images)}")
print(f"Valid images    : {len(valid_pairs)}")
print(f"Excluded       : {len(invalid_pairs)}")
print(f"Train          : {len(train_items)}")
print(f"Valid          : {len(valid_items)}")
print(f"Test           : {len(test_items)}")
print(f"Final total    : {final_total}")

print("\nOutput:")
print(OUTPUT_ROOT)
```

```python

# ============================================================
# CONFIGURATION
# ============================================================

DATASET_ROOT = Path(r"D:\ExDark_YOLO")

SPLITS = ["train", "valid", "test"]

CLASS_NAMES = [
    "Bicycle",
    "Boat",
    "Bottle",
    "Bus",
    "Car",
    "Cat",
    "Chair",
    "Cup",
    "Dog",
    "Motorbike",
    "People",
    "Table"
]

NUM_CLASSES = len(CLASS_NAMES)


# ============================================================
# VALIDATION FUNCTION
# ============================================================

def validate_split(split):

    image_dir = DATASET_ROOT / split / "images"
    label_dir = DATASET_ROOT / split / "labels"

    print("\n" + "=" * 80)
    print(f"{split.upper()} VALIDATION")
    print("=" * 80)

    # --------------------------------------------------------
    # Check directories
    # --------------------------------------------------------

    if not image_dir.exists():
        print(f"ERROR: Missing image directory: {image_dir}")
        return None

    if not label_dir.exists():
        print(f"ERROR: Missing label directory: {label_dir}")
        return None

    # --------------------------------------------------------
    # Find images
    # --------------------------------------------------------

    image_extensions = {
        ".jpg",
        ".jpeg",
        ".png",
        ".bmp",
        ".webp"
    }

    images = [
        p for p in image_dir.iterdir()
        if p.is_file()
        and p.suffix.lower() in image_extensions
    ]

    # --------------------------------------------------------
    # Find labels
    # --------------------------------------------------------

    labels = [
        p for p in label_dir.iterdir()
        if p.is_file()
        and p.suffix.lower() == ".txt"
    ]

    print(f"Images : {len(images)}")
    print(f"Labels : {len(labels)}")

    # --------------------------------------------------------
    # Match images and labels
    # --------------------------------------------------------

    image_stems = {
        p.stem.lower(): p
        for p in images
    }

    label_stems = {
        p.stem.lower(): p
        for p in labels
    }

    images_without_labels = sorted(
        set(image_stems) - set(label_stems)
    )

    labels_without_images = sorted(
        set(label_stems) - set(image_stems)
    )

    print(f"Images without labels : {len(images_without_labels)}")
    print(f"Labels without images : {len(labels_without_images)}")

    if images_without_labels:
        print("\nImages without labels:")

        for stem in images_without_labels[:20]:
            print("  ", image_stems[stem])

        if len(images_without_labels) > 20:
            print(
                f"  ... and "
                f"{len(images_without_labels) - 20} more"
            )

    if labels_without_images:
        print("\nLabels without images:")

        for stem in labels_without_images[:20]:
            print("  ", label_stems[stem])

        if len(labels_without_images) > 20:
            print(
                f"  ... and "
                f"{len(labels_without_images) - 20} more"
            )

    # --------------------------------------------------------
    # Annotation validation
    # --------------------------------------------------------

    invalid_lines = []
    invalid_class_ids = []
    invalid_boxes = []
    empty_labels = []
    unreadable_images = []

    class_objects = Counter()
    class_images = Counter()

    total_objects = 0

    # --------------------------------------------------------
    # Validate every image + label
    # --------------------------------------------------------

    for image_path in images:

        stem = image_path.stem.lower()

        label_path = label_stems.get(stem)

        # No corresponding label
        if label_path is None:
            continue

        # ----------------------------------------------------
        # Read image
        # ----------------------------------------------------

        try:

            with Image.open(image_path) as img:
                img_width, img_height = img.size

        except Exception as e:

            unreadable_images.append(
                (image_path, str(e))
            )

            continue

        # ----------------------------------------------------
        # Read label file
        # ----------------------------------------------------

        try:

            with open(
                label_path,
                "r",
                encoding="utf-8"
            ) as f:

                lines = f.readlines()

        except Exception as e:

            invalid_lines.append(
                (label_path, 0, f"Cannot read file: {e}")
            )

            continue

        # ----------------------------------------------------
        # Empty label
        # ----------------------------------------------------

        valid_objects_in_image = 0

        if not lines or not any(line.strip() for line in lines):

            empty_labels.append(label_path)
            continue

        # ----------------------------------------------------
        # Validate each YOLO line
        # ----------------------------------------------------

        for line_number, line in enumerate(
            lines,
            start=1
        ):

            line = line.strip()

            if not line:
                continue

            parts = line.split()

            # YOLO format requires exactly:
            #
            # class_id x_center y_center width height
            #

            if len(parts) != 5:

                invalid_lines.append(
                    (
                        label_path,
                        line_number,
                        line
                    )
                )

                continue

            try:

                class_id = int(parts[0])

                x_center = float(parts[1])
                y_center = float(parts[2])
                width = float(parts[3])
                height = float(parts[4])

            except ValueError:

                invalid_lines.append(
                    (
                        label_path,
                        line_number,
                        line
                    )
                )

                continue

            # ------------------------------------------------
            # Class ID validation
            # ------------------------------------------------

            if not (0 <= class_id < NUM_CLASSES):

                invalid_class_ids.append(
                    (
                        label_path,
                        line_number,
                        class_id
                    )
                )

                continue

            # ------------------------------------------------
            # Bounding box validation
            # ------------------------------------------------

            valid_box = (
                0 <= x_center <= 1
                and
                0 <= y_center <= 1
                and
                0 < width <= 1
                and
                0 < height <= 1
            )

            if not valid_box:

                invalid_boxes.append(
                    (
                        label_path,
                        line_number,
                        x_center,
                        y_center,
                        width,
                        height
                    )
                )

                continue

            # ------------------------------------------------
            # Check that box stays inside image
            # ------------------------------------------------

            x1 = x_center - width / 2
            y1 = y_center - height / 2
            x2 = x_center + width / 2
            y2 = y_center + height / 2

            if (
                x1 < 0
                or y1 < 0
                or x2 > 1
                or y2 > 1
            ):

                invalid_boxes.append(
                    (
                        label_path,
                        line_number,
                        x_center,
                        y_center,
                        width,
                        height
                    )
                )

                continue

            # ------------------------------------------------
            # Valid object
            # ------------------------------------------------

            class_objects[class_id] += 1
            valid_objects_in_image += 1
            total_objects += 1

        # ----------------------------------------------------
        # Count image once per class
        # ----------------------------------------------------

        if valid_objects_in_image > 0:

            # Read valid class IDs again for image count
            classes_in_image = set()

            for line in lines:

                parts = line.strip().split()

                if len(parts) != 5:
                    continue

                try:
                    class_id = int(parts[0])
                except ValueError:
                    continue

                if not (0 <= class_id < NUM_CLASSES):
                    continue

                try:
                    x = float(parts[1])
                    y = float(parts[2])
                    w = float(parts[3])
                    h = float(parts[4])
                except ValueError:
                    continue

                if not (
                    0 <= x <= 1
                    and 0 <= y <= 1
                    and 0 < w <= 1
                    and 0 < h <= 1
                ):
                    continue

                x1 = x - w / 2
                y1 = y - h / 2
                x2 = x + w / 2
                y2 = y + h / 2

                if (
                    x1 >= 0
                    and y1 >= 0
                    and x2 <= 1
                    and y2 <= 1
                ):
                    classes_in_image.add(class_id)

            for class_id in classes_in_image:
                class_images[class_id] += 1

    # ========================================================
    # PRINT RESULTS
    # ========================================================

    print("\n" + "-" * 80)
    print("ANNOTATION VALIDATION")
    print("-" * 80)

    print(f"Total objects       : {total_objects}")
    print(f"Invalid lines       : {len(invalid_lines)}")
    print(f"Invalid class IDs   : {len(invalid_class_ids)}")
    print(f"Invalid boxes       : {len(invalid_boxes)}")
    print(f"Empty label files   : {len(empty_labels)}")
    print(f"Unreadable images   : {len(unreadable_images)}")

    # --------------------------------------------------------
    # Class distribution
    # --------------------------------------------------------

    print("\n" + "-" * 80)
    print(f"{split.upper()} CLASS DISTRIBUTION")
    print("-" * 80)

    print(
        f"{'ID':<5}"
        f"{'Class':<15}"
        f"{'Objects':<12}"
        f"{'Images':<12}"
        f"{'Object %':<10}"
    )

    print("-" * 80)

    for class_id, class_name in enumerate(CLASS_NAMES):

        objects = class_objects[class_id]
        image_count = class_images[class_id]

        percentage = (
            objects / total_objects * 100
            if total_objects > 0
            else 0
        )

        print(
            f"{class_id:<5}"
            f"{class_name:<15}"
            f"{objects:<12}"
            f"{image_count:<12}"
            f"{percentage:.2f}%"
        )

    return {
        "images": len(images),
        "labels": len(labels),
        "images_without_labels": len(images_without_labels),
        "labels_without_images": len(labels_without_images),
        "objects": total_objects,
        "invalid_lines": len(invalid_lines),
        "invalid_class_ids": len(invalid_class_ids),
        "invalid_boxes": len(invalid_boxes),
        "empty_labels": len(empty_labels),
        "unreadable_images": len(unreadable_images),
        "class_objects": class_objects,
        "class_images": class_images
    }


# ============================================================
# RUN VALIDATION
# ============================================================

results = {}

for split in SPLITS:

    results[split] = validate_split(split)


# ============================================================
# FINAL DATASET SUMMARY
# ============================================================

print("\n\n" + "=" * 80)
print("FINAL EXDARK DATASET VALIDATION")
print("=" * 80)

total_images = 0
total_labels = 0
total_objects = 0

total_missing_labels = 0
total_extra_labels = 0
total_invalid_lines = 0
total_invalid_classes = 0
total_invalid_boxes = 0
total_empty_labels = 0
total_unreadable = 0

for split in SPLITS:

    r = results[split]

    total_images += r["images"]
    total_labels += r["labels"]
    total_objects += r["objects"]

    total_missing_labels += r["images_without_labels"]
    total_extra_labels += r["labels_without_images"]

    total_invalid_lines += r["invalid_lines"]
    total_invalid_classes += r["invalid_class_ids"]
    total_invalid_boxes += r["invalid_boxes"]
    total_empty_labels += r["empty_labels"]
    total_unreadable += r["unreadable_images"]


print(f"\nImages                 : {total_images}")
print(f"Labels                 : {total_labels}")
print(f"Total objects          : {total_objects}")

print(f"\nImages without labels  : {total_missing_labels}")
print(f"Labels without images  : {total_extra_labels}")

print(f"\nInvalid annotation lines : {total_invalid_lines}")
print(f"Invalid class IDs       : {total_invalid_classes}")
print(f"Invalid bounding boxes  : {total_invalid_boxes}")
print(f"Empty label files       : {total_empty_labels}")
print(f"Unreadable images       : {total_unreadable}")


# ============================================================
# OVERALL CLASS DISTRIBUTION
# ============================================================

overall_objects = Counter()
overall_images = Counter()

for split in SPLITS:

    overall_objects.update(
        results[split]["class_objects"]
    )

    overall_images.update(
        results[split]["class_images"]
    )


print("\n" + "=" * 80)
print("OVERALL CLASS DISTRIBUTION")
print("=" * 80)

print(
    f"{'ID':<5}"
    f"{'Class':<15}"
    f"{'Objects':<12}"
    f"{'Images':<12}"
    f"{'Object %':<10}"
)

print("-" * 80)

for class_id, class_name in enumerate(CLASS_NAMES):

    objects = overall_objects[class_id]
    image_count = overall_images[class_id]

    percentage = (
        objects / total_objects * 100
        if total_objects > 0
        else 0
    )

    print(
        f"{class_id:<5}"
        f"{class_name:<15}"
        f"{objects:<12}"
        f"{image_count:<12}"
        f"{percentage:.2f}%"
    )


# ============================================================
# FINAL PASS / FAIL
# ============================================================

print("\n" + "=" * 80)
print("FINAL STATUS")
print("=" * 80)

checks = {
    "Expected total images = 7362":
        total_images == 7362,

    "Images = labels":
        total_images == total_labels,

    "No images without labels":
        total_missing_labels == 0,

    "No labels without images":
        total_extra_labels == 0,

    "No invalid annotation lines":
        total_invalid_lines == 0,

    "No invalid class IDs":
        total_invalid_classes == 0,

    "No invalid bounding boxes":
        total_invalid_boxes == 0,

    "No empty label files":
        total_empty_labels == 0,

    "No unreadable images":
        total_unreadable == 0
}


all_passed = True

for check, passed in checks.items():

    if passed:
        print(f"PASS  - {check}")

    else:
        print(f"FAIL  - {check}")
        all_passed = False


print("\n" + "=" * 80)

if all_passed:
    print("DATASET VALIDATION PASSED")
    print("The ExDark YOLO dataset is ready for the next step.")
else:
    print("DATASET VALIDATION NEEDS INVESTIGATION")

print("=" * 80)
```

```python


# ============================================================
# 1. PATHS
# ============================================================

IMAGE_ROOT = Path(r"D:\ExDark\ExDark")
ANNOTATION_ROOT = Path(r"D:\ExDark_Anno\ExDark_Anno")
OUTPUT_ROOT = Path(r"D:\ExDark_YOLO")


# ============================================================
# 2. CLASS DEFINITIONS
# ============================================================

CLASS_NAMES = [
    "Bicycle",
    "Boat",
    "Bottle",
    "Bus",
    "Car",
    "Cat",
    "Chair",
    "Cup",
    "Dog",
    "Motorbike",
    "People",
    "Table"
]

CLASS_TO_ID = {
    name.lower(): i
    for i, name in enumerate(CLASS_NAMES)
}


# ============================================================
# 3. RECREATE OUTPUT DIRECTORY
# ============================================================

if OUTPUT_ROOT.exists():

    print("=" * 80)
    print("REMOVING PREVIOUS YOLO DATASET")
    print("=" * 80)

    shutil.rmtree(OUTPUT_ROOT)

    print("Previous dataset removed.")


# Create directories

for split in ["train", "valid", "test"]:

    (OUTPUT_ROOT / split / "images").mkdir(
        parents=True,
        exist_ok=True
    )

    (OUTPUT_ROOT / split / "labels").mkdir(
        parents=True,
        exist_ok=True
    )


# ============================================================
# 4. FIND ORIGINAL IMAGES
# ============================================================

IMAGE_EXTENSIONS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".webp"
}

images = []

for path in IMAGE_ROOT.rglob("*"):

    if "__MACOSX" in path.parts:
        continue

    if (
        path.is_file()
        and path.suffix.lower() in IMAGE_EXTENSIONS
    ):
        images.append(path)


print("\n" + "=" * 80)
print("EXDARK DATASET CONVERSION")
print("=" * 80)

print(f"Image root       : {IMAGE_ROOT}")
print(f"Annotation root  : {ANNOTATION_ROOT}")
print(f"Output root      : {OUTPUT_ROOT}")
print(f"Original images  : {len(images)}")


# ============================================================
# 5. INDEX ANNOTATIONS
# ============================================================

annotation_index = {}

for path in ANNOTATION_ROOT.rglob("*.txt"):

    if "__MACOSX" in path.parts:
        continue

    # Example:
    #
    # 2015_04089.jpg.txt
    #
    # Matching is case-insensitive.
    key = path.name.lower()

    annotation_index[key] = path


print(f"Annotation files : {len(annotation_index)}")


# ============================================================
# 6. PARSE EXDARK ANNOTATION
# ============================================================

def parse_annotation(
    annotation_path,
    img_width,
    img_height
):

    yolo_annotations = []

    with open(
        annotation_path,
        "r",
        encoding="utf-8",
        errors="ignore"
    ) as f:

        lines = f.readlines()


    for line in lines:

        line = line.strip()

        # Skip empty lines
        if not line:
            continue

        # Skip ExDark header
        if line.startswith("%"):
            continue


        parts = line.split()

        # Expected:
        #
        # class left top width height ...
        #

        if len(parts) < 5:
            continue


        class_name = parts[0].strip()

        class_id = CLASS_TO_ID.get(
            class_name.lower()
        )

        if class_id is None:
            continue


        try:

            left = float(parts[1])
            top = float(parts[2])
            width = float(parts[3])
            height = float(parts[4])

        except ValueError:

            continue


        # ----------------------------------------------------
        # Original ExDark format
        #
        # left, top, width, height
        # ----------------------------------------------------

        x1 = left
        y1 = top

        x2 = left + width
        y2 = top + height


        # ----------------------------------------------------
        # CLIP TO IMAGE BOUNDARIES
        #
        # This is important because some ExDark annotations
        # extend slightly outside the image.
        # ----------------------------------------------------

        x1 = max(
            0.0,
            min(x1, float(img_width))
        )

        y1 = max(
            0.0,
            min(y1, float(img_height))
        )

        x2 = max(
            0.0,
            min(x2, float(img_width))
        )

        y2 = max(
            0.0,
            min(y2, float(img_height))
        )


        # ----------------------------------------------------
        # Ignore completely invalid boxes
        # ----------------------------------------------------

        if x2 <= x1 or y2 <= y1:
            continue


        # ----------------------------------------------------
        # Calculate clipped box
        # ----------------------------------------------------

        box_width = x2 - x1
        box_height = y2 - y1


        # ----------------------------------------------------
        # YOLO center
        # ----------------------------------------------------

        x_center = (x1 + x2) / 2.0
        y_center = (y1 + y2) / 2.0


        # ----------------------------------------------------
        # Normalize
        # ----------------------------------------------------

        x_center /= img_width
        y_center /= img_height

        box_width /= img_width
        box_height /= img_height


        # ----------------------------------------------------
        # Final sanity check on YOLO values
        #
        # We validate the YOLO representation itself.
        # We do NOT reconstruct x2/y2 and reject boxes
        # because of tiny floating-point/rounding effects.
        # ----------------------------------------------------

        if not (
            0.0 <= x_center <= 1.0
            and
            0.0 <= y_center <= 1.0
            and
            0.0 < box_width <= 1.0
            and
            0.0 < box_height <= 1.0
        ):

            continue


        # ----------------------------------------------------
        # Write YOLO line
        # ----------------------------------------------------

        yolo_line = (
            f"{class_id} "
            f"{x_center:.6f} "
            f"{y_center:.6f} "
            f"{box_width:.6f} "
            f"{box_height:.6f}"
        )

        yolo_annotations.append(
            yolo_line
        )


    return yolo_annotations


# ============================================================
# 7. MATCH + CONVERT
# ============================================================

valid_pairs = []
invalid_pairs = []

print("\n" + "=" * 80)
print("MATCHING AND CONVERTING")
print("=" * 80)


for i, image_path in enumerate(images, 1):

    # --------------------------------------------------------
    # Find annotation
    #
    # Image:
    # 2015_04089.jpg
    #
    # Annotation:
    # 2015_04089.jpg.txt
    # --------------------------------------------------------

    annotation_name = (
        image_path.name.lower() + ".txt"
    )

    annotation_path = annotation_index.get(
        annotation_name
    )


    if annotation_path is None:

        invalid_pairs.append(
            (
                image_path,
                None,
                "Annotation not found"
            )
        )

        continue


    # --------------------------------------------------------
    # Read image dimensions
    # --------------------------------------------------------

    try:

        with Image.open(image_path) as img:

            img_width, img_height = img.size

    except Exception as e:

        invalid_pairs.append(
            (
                image_path,
                annotation_path,
                f"Could not read image: {e}"
            )
        )

        continue


    # --------------------------------------------------------
    # Parse annotations
    # --------------------------------------------------------

    yolo_annotations = parse_annotation(
        annotation_path,
        img_width,
        img_height
    )


    # --------------------------------------------------------
    # Exclude images with no usable objects
    # --------------------------------------------------------

    if not yolo_annotations:

        invalid_pairs.append(
            (
                image_path,
                annotation_path,
                "No valid annotations found"
            )
        )

        continue


    # --------------------------------------------------------
    # Keep valid image/annotation pair
    # --------------------------------------------------------

    valid_pairs.append(
        (
            image_path,
            annotation_path,
            yolo_annotations
        )
    )


    # Progress
    if i % 500 == 0:

        print(
            f"Processed {i}/{len(images)} images..."
        )


# ============================================================
# 8. CONVERSION SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("CONVERSION SUMMARY")
print("=" * 80)

print(f"Original images      : {len(images)}")
print(f"Valid image pairs    : {len(valid_pairs)}")
print(f"Invalid/excluded     : {len(invalid_pairs)}")


# ============================================================
# 9. DISPLAY EXCLUDED SAMPLES
# ============================================================

if invalid_pairs:

    print("\n" + "=" * 80)
    print("EXCLUDED SAMPLES")
    print("=" * 80)

    for i, (
        image_path,
        annotation_path,
        reason
    ) in enumerate(invalid_pairs, 1):

        print(f"\n{i}. IMAGE:")
        print(image_path)

        print("   ANNOTATION:")
        print(annotation_path)

        print("   REASON:")
        print(reason)


# ============================================================
# 10. STRATIFIED 70/15/15 SPLIT
# ============================================================

random.seed(42)

groups = defaultdict(list)


for item in valid_pairs:

    image_path = item[0]

    # Original ExDark category folder
    original_class = image_path.parent.name

    groups[original_class].append(item)


train_items = []
valid_items = []
test_items = []


for class_name, items in groups.items():

    random.shuffle(items)

    n = len(items)

    train_count = int(n * 0.70)

    valid_count = int(n * 0.15)

    train_items.extend(
        items[:train_count]
    )

    valid_items.extend(
        items[
            train_count:
            train_count + valid_count
        ]
    )

    test_items.extend(
        items[
            train_count + valid_count:
        ]
    )


# Shuffle final splits

random.shuffle(train_items)
random.shuffle(valid_items)
random.shuffle(test_items)


# ============================================================
# 11. SPLIT SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("DATASET SPLIT")
print("=" * 80)

print(f"Train : {len(train_items)}")
print(f"Valid : {len(valid_items)}")
print(f"Test  : {len(test_items)}")

split_total = (
    len(train_items)
    + len(valid_items)
    + len(test_items)
)

print(f"Total : {split_total}")


# ============================================================
# 12. COPY IMAGES + WRITE LABELS
# ============================================================

def write_split(
    items,
    image_dir,
    label_dir
):

    for (
        image_path,
        annotation_path,
        yolo_annotations
    ) in items:

        # ----------------------------------------------------
        # Preserve original image filename
        # ----------------------------------------------------

        image_destination = (
            image_dir / image_path.name
        )


        # ----------------------------------------------------
        # Correct YOLO label name
        #
        # 2015_04089.jpg
        #
        # becomes:
        #
        # 2015_04089.txt
        # ----------------------------------------------------

        label_destination = (
            label_dir / f"{image_path.stem}.txt"
        )


        # ----------------------------------------------------
        # Copy original image unchanged
        # ----------------------------------------------------

        shutil.copy2(
            image_path,
            image_destination
        )


        # ----------------------------------------------------
        # Write YOLO labels
        # ----------------------------------------------------

        with open(
            label_destination,
            "w",
            encoding="utf-8"
        ) as f:

            f.write(
                "\n".join(yolo_annotations)
            )


# ============================================================
# 13. WRITE ALL SPLITS
# ============================================================

print("\n" + "=" * 80)
print("WRITING DATASET")
print("=" * 80)


write_split(
    train_items,
    OUTPUT_ROOT / "train" / "images",
    OUTPUT_ROOT / "train" / "labels"
)


write_split(
    valid_items,
    OUTPUT_ROOT / "valid" / "images",
    OUTPUT_ROOT / "valid" / "labels"
)


write_split(
    test_items,
    OUTPUT_ROOT / "test" / "images",
    OUTPUT_ROOT / "test" / "labels"
)


print("Dataset files written successfully.")


# ============================================================
# 14. CREATE data.yaml
# ============================================================

yaml_path = OUTPUT_ROOT / "data.yaml"


yaml_text = f"""path: {OUTPUT_ROOT.as_posix()}
train: train/images
val: valid/images
test: test/images

nc: {len(CLASS_NAMES)}

names:
"""


for i, class_name in enumerate(CLASS_NAMES):

    yaml_text += (
        f"  {i}: {class_name}\n"
    )


with open(
    yaml_path,
    "w",
    encoding="utf-8"
) as f:

    f.write(yaml_text)


print("\n" + "=" * 80)
print("DATA.YAML")
print("=" * 80)

print(yaml_text)


# ============================================================
# 15. FINAL RESULT
# ============================================================

print("=" * 80)
print("FINAL CONVERSION RESULT")
print("=" * 80)

print(f"Original images : {len(images)}")
print(f"Valid images    : {len(valid_pairs)}")
print(f"Excluded        : {len(invalid_pairs)}")

print(f"\nTrain           : {len(train_items)}")
print(f"Valid           : {len(valid_items)}")
print(f"Test            : {len(test_items)}")
print(f"Final total     : {split_total}")

print("\nOutput directory:")
print(OUTPUT_ROOT)

print("\n" + "=" * 80)

if (
    len(valid_pairs) == 7362
    and
    len(invalid_pairs) == 1
    and
    split_total == 7362
):

    print("CONVERSION PASSED")
    print("7362 usable ExDark images were converted successfully.")

else:

    print("CHECK CONVERSION RESULTS")

print("=" * 80)
```

```python


# ============================================================
# FINAL EXDARK YOLO DATASET VALIDATION
# ============================================================

DATASET = Path(r"D:\ExDark_YOLO")

SPLITS = ["train", "valid", "test"]

CLASS_NAMES = [
    "Bicycle",
    "Boat",
    "Bottle",
    "Bus",
    "Car",
    "Cat",
    "Chair",
    "Cup",
    "Dog",
    "Motorbike",
    "People",
    "Table"
]

NUM_CLASSES = len(CLASS_NAMES)
EXPECTED_TOTAL = 7362

IMAGE_EXTENSIONS = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}

# ------------------------------------------------------------
# Helper
# ------------------------------------------------------------

def print_line(char="=", n=90):
    print(char * n)


# ------------------------------------------------------------
# Initial checks
# ------------------------------------------------------------

print_line()
print("FINAL EXDARK YOLO DATASET VALIDATION")
print_line()

print(f"Dataset: {DATASET}")
print(f"Expected usable images: {EXPECTED_TOTAL}")
print(f"Expected classes: {NUM_CLASSES}")
print()

if not DATASET.exists():
    raise FileNotFoundError(f"Dataset not found: {DATASET}")

# ------------------------------------------------------------
# Check data.yaml
# ------------------------------------------------------------

yaml_path = DATASET / "data.yaml"

print("CHECKING data.yaml")
print_line("-", 90)

if not yaml_path.exists():
    print("❌ data.yaml is missing")
else:
    try:
        with open(yaml_path, "r", encoding="utf-8") as f:
            data = yaml.safe_load(f)

        print("✓ data.yaml found")

        if data.get("nc") == NUM_CLASSES:
            print(f"✓ Number of classes: {data.get('nc')}")
        else:
            print(f"❌ Wrong nc: {data.get('nc')}")

        yaml_names = data.get("names", [])

        if isinstance(yaml_names, dict):
            yaml_names = [yaml_names[i] for i in range(NUM_CLASSES)]

        if yaml_names == CLASS_NAMES:
            print("✓ Class names are correct")
        else:
            print("❌ Class names do not match")
            print("YAML:", yaml_names)

    except Exception as e:
        print(f"❌ Could not read data.yaml: {e}")

print()

# ------------------------------------------------------------
# Global counters
# ------------------------------------------------------------

grand_images = 0
grand_labels = 0
grand_objects = 0

global_class_objects = Counter()
global_class_images = Counter()

total_invalid_lines = 0
total_invalid_class_ids = 0
total_invalid_values = 0
total_empty_labels = 0
total_unreadable_images = 0

all_global_image_stems = set()
all_global_label_stems = set()

# ------------------------------------------------------------
# Validate each split
# ------------------------------------------------------------

for split in SPLITS:

    print_line()
    print(f"{split.upper()} VALIDATION")
    print_line()

    images_dir = DATASET / split / "images"
    labels_dir = DATASET / split / "labels"

    if not images_dir.exists():
        print(f"❌ Missing images directory: {images_dir}")
        continue

    if not labels_dir.exists():
        print(f"❌ Missing labels directory: {labels_dir}")
        continue

    # --------------------------------------------------------
    # Find files
    # --------------------------------------------------------

    images = [
        p for p in images_dir.iterdir()
        if p.is_file() and p.suffix.lower() in IMAGE_EXTENSIONS
    ]

    labels = [
        p for p in labels_dir.iterdir()
        if p.is_file() and p.suffix.lower() == ".txt"
    ]

    image_stems = {p.stem for p in images}
    label_stems = {p.stem for p in labels}

    grand_images += len(images)
    grand_labels += len(labels)

    all_global_image_stems.update(
        f"{split}/{stem}" for stem in image_stems
    )

    all_global_label_stems.update(
        f"{split}/{stem}" for stem in label_stems
    )

    print(f"Images: {len(images)}")
    print(f"Labels: {len(labels)}")

    # --------------------------------------------------------
    # Image-label matching
    # --------------------------------------------------------

    images_without_labels = image_stems - label_stems
    labels_without_images = label_stems - image_stems

    print(f"Images without labels: {len(images_without_labels)}")
    print(f"Labels without images: {len(labels_without_images)}")

    if images_without_labels:
        print("\nExamples of images without labels:")
        for x in sorted(images_without_labels)[:10]:
            print("  ", x)

    if labels_without_images:
        print("\nExamples of labels without images:")
        for x in sorted(labels_without_images)[:10]:
            print("  ", x)

    # --------------------------------------------------------
    # Class counters for this split
    # --------------------------------------------------------

    split_class_objects = Counter()
    split_class_images = Counter()

    split_invalid_lines = 0
    split_invalid_class_ids = 0
    split_invalid_values = 0
    split_empty_labels = 0
    split_unreadable_images = 0

    # --------------------------------------------------------
    # Validate images
    # --------------------------------------------------------

    print()
    print("IMAGE VALIDATION")
    print_line("-", 90)

    for image_path in images:

        try:
            with Image.open(image_path) as img:
                img.verify()

        except Exception:
            split_unreadable_images += 1
            print(f"❌ Unreadable image: {image_path}")

    print(f"Unreadable images: {split_unreadable_images}")

    # --------------------------------------------------------
    # Validate labels
    # --------------------------------------------------------

    print()
    print("ANNOTATION VALIDATION")
    print_line("-", 90)

    for label_path in labels:

        try:
            with open(label_path, "r", encoding="utf-8") as f:
                lines = [line.strip() for line in f if line.strip()]

        except Exception as e:
            print(f"❌ Could not read: {label_path}")
            split_invalid_lines += 1
            continue

        # Empty label file
        if len(lines) == 0:
            split_empty_labels += 1
            print(f"❌ Empty label file: {label_path.name}")
            continue

        image_has_valid_object = False

        for line_number, line in enumerate(lines, start=1):

            parts = line.split()

            # ------------------------------------------------
            # YOLO format must contain exactly 5 values
            # ------------------------------------------------

            if len(parts) != 5:
                split_invalid_lines += 1

                if split_invalid_lines <= 20:
                    print(
                        f"❌ Invalid format: "
                        f"{label_path.name}, line {line_number}: {line}"
                    )

                continue

            try:
                class_id = int(parts[0])

                x_center = float(parts[1])
                y_center = float(parts[2])
                width = float(parts[3])
                height = float(parts[4])

            except ValueError:

                split_invalid_values += 1

                if split_invalid_values <= 20:
                    print(
                        f"❌ Non-numeric values: "
                        f"{label_path.name}, line {line_number}: {line}"
                    )

                continue

            # ------------------------------------------------
            # Validate class ID
            # ------------------------------------------------

            if not (0 <= class_id < NUM_CLASSES):

                split_invalid_class_ids += 1

                if split_invalid_class_ids <= 20:
                    print(
                        f"❌ Invalid class ID: "
                        f"{label_path.name}, line {line_number}: "
                        f"{class_id}"
                    )

                continue

            # ------------------------------------------------
            # Validate YOLO values directly
            #
            # IMPORTANT:
            # We do NOT reconstruct x1/x2/y1/y2 here.
            # The conversion already clipped the original
            # bounding boxes to the image boundaries.
            # ------------------------------------------------

            valid_box = (
                0.0 <= x_center <= 1.0
                and 0.0 <= y_center <= 1.0
                and 0.0 < width <= 1.0
                and 0.0 < height <= 1.0
            )

            if not valid_box:

                split_invalid_values += 1

                if split_invalid_values <= 20:
                    print(
                        f"❌ Invalid YOLO values: "
                        f"{label_path.name}, line {line_number}: {line}"
                    )

                continue

            # ------------------------------------------------
            # Valid object
            # ------------------------------------------------

            split_class_objects[class_id] += 1
            global_class_objects[class_id] += 1

            image_has_valid_object = True

        if image_has_valid_object:
            # Count this image once for each class present
            classes_in_image = set()

            for line in lines:
                parts = line.split()

                if len(parts) != 5:
                    continue

                try:
                    class_id = int(parts[0])
                    x_center = float(parts[1])
                    y_center = float(parts[2])
                    width = float(parts[3])
                    height = float(parts[4])

                    if (
                        0 <= class_id < NUM_CLASSES
                        and 0.0 <= x_center <= 1.0
                        and 0.0 <= y_center <= 1.0
                        and 0.0 < width <= 1.0
                        and 0.0 < height <= 1.0
                    ):
                        classes_in_image.add(class_id)

                except ValueError:
                    continue

            for class_id in classes_in_image:
                split_class_images[class_id] += 1
                global_class_images[class_id] += 1

    # --------------------------------------------------------
    # Split summary
    # --------------------------------------------------------

    split_objects = sum(split_class_objects.values())

    grand_objects += split_objects

    total_invalid_lines += split_invalid_lines
    total_invalid_class_ids += split_invalid_class_ids
    total_invalid_values += split_invalid_values
    total_empty_labels += split_empty_labels
    total_unreadable_images += split_unreadable_images

    print()
    print("SPLIT SUMMARY")
    print_line("-", 90)

    print(f"Images: {len(images)}")
    print(f"Labels: {len(labels)}")
    print(f"Objects: {split_objects}")
    print(f"Invalid annotation lines: {split_invalid_lines}")
    print(f"Invalid class IDs: {split_invalid_class_ids}")
    print(f"Invalid numeric/YOLO values: {split_invalid_values}")
    print(f"Empty label files: {split_empty_labels}")
    print(f"Unreadable images: {split_unreadable_images}")

    # --------------------------------------------------------
    # Class distribution
    # --------------------------------------------------------

    print()
    print(f"{split.upper()} CLASS DISTRIBUTION")
    print_line("-", 90)

    print(
        f"{'ID':<4}"
        f"{'Class':<15}"
        f"{'Objects':>12}"
        f"{'Images':>12}"
    )

    print_line("-", 90)

    for class_id, class_name in enumerate(CLASS_NAMES):

        print(
            f"{class_id:<4}"
            f"{class_name:<15}"
            f"{split_class_objects[class_id]:>12}"
            f"{split_class_images[class_id]:>12}"
        )

# ============================================================
# GLOBAL SUMMARY
# ============================================================

print()
print_line()
print("FINAL DATASET SUMMARY")
print_line()

print(f"Total images:              {grand_images}")
print(f"Total labels:              {grand_labels}")
print(f"Total objects:             {grand_objects}")
print(f"Expected images:           {EXPECTED_TOTAL}")
print()

print(f"Invalid annotation lines:  {total_invalid_lines}")
print(f"Invalid class IDs:         {total_invalid_class_ids}")
print(f"Invalid YOLO values:       {total_invalid_values}")
print(f"Empty label files:         {total_empty_labels}")
print(f"Unreadable images:         {total_unreadable_images}")

# ============================================================
# FINAL CLASS DISTRIBUTION
# ============================================================

print()
print_line()
print("FINAL CLASS DISTRIBUTION")
print_line()

print(
    f"{'ID':<4}"
    f"{'Class':<15}"
    f"{'Objects':>12}"
    f"{'Images':>12}"
)

print_line("-", 90)

for class_id, class_name in enumerate(CLASS_NAMES):

    print(
        f"{class_id:<4}"
        f"{class_name:<15}"
        f"{global_class_objects[class_id]:>12}"
        f"{global_class_images[class_id]:>12}"
    )

# ============================================================
# FINAL PASS / FAIL
# ============================================================

print()
print_line()
print("FINAL VALIDATION RESULT")
print_line()

errors = []

if grand_images != EXPECTED_TOTAL:
    errors.append(
        f"Expected {EXPECTED_TOTAL} images, found {grand_images}"
    )

if grand_labels != grand_images:
    errors.append(
        f"Image/label count mismatch: "
        f"{grand_images} images vs {grand_labels} labels"
    )

if total_invalid_lines > 0:
    errors.append(
        f"{total_invalid_lines} invalid annotation lines"
    )

if total_invalid_class_ids > 0:
    errors.append(
        f"{total_invalid_class_ids} invalid class IDs"
    )

if total_invalid_values > 0:
    errors.append(
        f"{total_invalid_values} invalid YOLO values"
    )

if total_empty_labels > 0:
    errors.append(
        f"{total_empty_labels} empty label files"
    )

if total_unreadable_images > 0:
    errors.append(
        f"{total_unreadable_images} unreadable images"
    )

if all_global_image_stems != all_global_label_stems:
    errors.append("Image/label filename mismatch exists")

if errors:

    print("❌ VALIDATION FAILED")
    print()

    for error in errors:
        print(" -", error)

else:

    print("✅ VALIDATION PASSED")
    print()
    print(f"✓ {grand_images} images")
    print(f"✓ {grand_labels} labels")
    print(f"✓ {grand_objects} objects")
    print("✓ All class IDs valid")
    print("✓ All YOLO values valid")
    print("✓ No empty labels")
    print("✓ All images readable")
    print("✓ All images have matching labels")
    print("✓ All labels have matching images")
    print("✓ Train + valid + test = 7,362 usable images")

print()
print_line()
print("VALIDATION COMPLETE")
print_line()
```

```python

# ============================================================
# FINAL EXDARK YOLO DATASET VALIDATION
# ============================================================

DATASET = Path(r"D:\ExDark_YOLO")

SPLITS = ["train", "valid", "test"]

CLASS_NAMES = [
    "Bicycle",
    "Boat",
    "Bottle",
    "Bus",
    "Car",
    "Cat",
    "Chair",
    "Cup",
    "Dog",
    "Motorbike",
    "People",
    "Table"
]

NUM_CLASSES = 12
EXPECTED_TOTAL = 7362

IMAGE_EXTENSIONS = {
    ".jpg", ".jpeg", ".png", ".bmp", ".webp"
}

# ============================================================
# INITIAL CHECK
# ============================================================

print("=" * 100)
print("FINAL EXDARK YOLO DATASET VALIDATION")
print("=" * 100)

print(f"Dataset: {DATASET}")
print(f"Expected usable images: {EXPECTED_TOTAL}")
print(f"Expected classes: {NUM_CLASSES}")
print()

if not DATASET.exists():
    raise FileNotFoundError(
        f"Dataset does not exist: {DATASET}"
    )

# ============================================================
# DATA.YAML CHECK
# ============================================================

print("=" * 100)
print("DATA.YAML CHECK")
print("=" * 100)

yaml_path = DATASET / "data.yaml"

if not yaml_path.exists():

    print("❌ data.yaml is missing")

else:

    print("✓ data.yaml found")
    print()

    yaml_text = yaml_path.read_text(
        encoding="utf-8"
    )

    print(yaml_text)

print()

# ============================================================
# GLOBAL COUNTERS
# ============================================================

total_images = 0
total_labels = 0
total_objects = 0

total_invalid_lines = 0
total_invalid_class_ids = 0
total_invalid_values = 0
total_empty_labels = 0
total_unreadable_images = 0

global_class_objects = Counter()
global_class_images = Counter()

# ============================================================
# SPLIT VALIDATION
# ============================================================

for split in SPLITS:

    print("=" * 100)
    print(f"{split.upper()} VALIDATION")
    print("=" * 100)

    images_dir = DATASET / split / "images"
    labels_dir = DATASET / split / "labels"

    # --------------------------------------------------------
    # Directory checks
    # --------------------------------------------------------

    if not images_dir.exists():

        print(f"❌ Missing directory: {images_dir}")
        continue

    if not labels_dir.exists():

        print(f"❌ Missing directory: {labels_dir}")
        continue

    # --------------------------------------------------------
    # Find images and labels
    # --------------------------------------------------------

    images = sorted([
        p for p in images_dir.iterdir()
        if p.is_file()
        and p.suffix.lower() in IMAGE_EXTENSIONS
    ])

    labels = sorted([
        p for p in labels_dir.iterdir()
        if p.is_file()
        and p.suffix.lower() == ".txt"
    ])

    image_stems = {p.stem for p in images}
    label_stems = {p.stem for p in labels}

    print(f"Images: {len(images)}")
    print(f"Labels: {len(labels)}")

    total_images += len(images)
    total_labels += len(labels)

    # --------------------------------------------------------
    # Image / label matching
    # --------------------------------------------------------

    images_without_labels = image_stems - label_stems
    labels_without_images = label_stems - image_stems

    print(
        f"Images without labels: "
        f"{len(images_without_labels)}"
    )

    print(
        f"Labels without images: "
        f"{len(labels_without_images)}"
    )

    if images_without_labels:

        print("\nExamples:")
        for stem in sorted(images_without_labels)[:10]:
            print("  ", stem)

    if labels_without_images:

        print("\nExamples:")
        for stem in sorted(labels_without_images)[:10]:
            print("  ", stem)

    # --------------------------------------------------------
    # Counters for this split
    # --------------------------------------------------------

    split_objects = 0

    split_invalid_lines = 0
    split_invalid_class_ids = 0
    split_invalid_values = 0
    split_empty_labels = 0
    split_unreadable_images = 0

    split_class_objects = Counter()
    split_class_images = Counter()

    # ========================================================
    # IMAGE VALIDATION
    # ========================================================

    print()
    print("-" * 100)
    print(f"{split.upper()} IMAGE VALIDATION")
    print("-" * 100)

    for image_path in images:

        try:

            with Image.open(image_path) as img:
                img.verify()

        except Exception:

            split_unreadable_images += 1

            if split_unreadable_images <= 10:
                print(
                    f"❌ Unreadable image: "
                    f"{image_path.name}"
                )

    print(
        f"Unreadable images: "
        f"{split_unreadable_images}"
    )

    # ========================================================
    # ANNOTATION VALIDATION
    # ========================================================

    print()
    print("-" * 100)
    print(f"{split.upper()} ANNOTATION VALIDATION")
    print("-" * 100)

    for label_path in labels:

        try:

            with open(
                label_path,
                "r",
                encoding="utf-8"
            ) as f:

                lines = [
                    line.strip()
                    for line in f
                    if line.strip()
                ]

        except Exception as e:

            split_invalid_lines += 1

            print(
                f"❌ Could not read "
                f"{label_path.name}: {e}"
            )

            continue

        # ----------------------------------------------------
        # Empty label
        # ----------------------------------------------------

        if len(lines) == 0:

            split_empty_labels += 1

            if split_empty_labels <= 20:
                print(
                    f"❌ Empty label: "
                    f"{label_path.name}"
                )

            continue

        classes_in_image = set()

        # ----------------------------------------------------
        # Validate every annotation
        # ----------------------------------------------------

        for line_number, line in enumerate(
            lines,
            start=1
        ):

            parts = line.split()

            # ------------------------------------------------
            # YOLO format
            # ------------------------------------------------

            if len(parts) != 5:

                split_invalid_lines += 1

                if split_invalid_lines <= 20:
                    print(
                        f"❌ Invalid format: "
                        f"{label_path.name}, "
                        f"line {line_number}"
                    )
                    print(
                        f"   {line}"
                    )

                continue

            # ------------------------------------------------
            # Convert values
            # ------------------------------------------------

            try:

                class_id = int(parts[0])

                x_center = float(parts[1])
                y_center = float(parts[2])
                width = float(parts[3])
                height = float(parts[4])

            except ValueError:

                split_invalid_values += 1

                if split_invalid_values <= 20:
                    print(
                        f"❌ Non-numeric values: "
                        f"{label_path.name}, "
                        f"line {line_number}"
                    )
                    print(
                        f"   {line}"
                    )

                continue

            # ------------------------------------------------
            # Class ID
            # ------------------------------------------------

            if not (
                0 <= class_id < NUM_CLASSES
            ):

                split_invalid_class_ids += 1

                if split_invalid_class_ids <= 20:
                    print(
                        f"❌ Invalid class ID: "
                        f"{label_path.name}, "
                        f"line {line_number}: "
                        f"{class_id}"
                    )

                continue

            # ------------------------------------------------
            # YOLO normalized values
            # ------------------------------------------------
            #
            # IMPORTANT:
            #
            # We validate the YOLO values directly.
            #
            # We DO NOT reconstruct x1/x2/y1/y2.
            #
            # This avoids false errors caused by the
            # six-decimal rounding used during conversion.
            #
            # ------------------------------------------------

            valid_box = (
                0.0 <= x_center <= 1.0
                and 0.0 <= y_center <= 1.0
                and 0.0 < width <= 1.0
                and 0.0 < height <= 1.0
            )

            if not valid_box:

                split_invalid_values += 1

                if split_invalid_values <= 20:
                    print(
                        f"❌ Invalid YOLO values: "
                        f"{label_path.name}, "
                        f"line {line_number}"
                    )
                    print(
                        f"   {line}"
                    )

                continue

            # ------------------------------------------------
            # Valid object
            # ------------------------------------------------

            split_objects += 1

            split_class_objects[class_id] += 1
            global_class_objects[class_id] += 1

            classes_in_image.add(class_id)

        # ----------------------------------------------------
        # Count image once per class
        # ----------------------------------------------------

        for class_id in classes_in_image:

            split_class_images[class_id] += 1
            global_class_images[class_id] += 1

    # ========================================================
    # SPLIT SUMMARY
    # ========================================================

    print()
    print("-" * 100)
    print(f"{split.upper()} SUMMARY")
    print("-" * 100)

    print(f"Images:                    {len(images)}")
    print(f"Labels:                    {len(labels)}")
    print(f"Objects:                   {split_objects}")

    print(
        f"Invalid annotation lines:  "
        f"{split_invalid_lines}"
    )

    print(
        f"Invalid class IDs:         "
        f"{split_invalid_class_ids}"
    )

    print(
        f"Invalid YOLO values:       "
        f"{split_invalid_values}"
    )

    print(
        f"Empty label files:         "
        f"{split_empty_labels}"
    )

    print(
        f"Unreadable images:         "
        f"{split_unreadable_images}"
    )

    # --------------------------------------------------------
    # Update global counters
    # --------------------------------------------------------

    total_objects += split_objects

    total_invalid_lines += split_invalid_lines
    total_invalid_class_ids += split_invalid_class_ids
    total_invalid_values += split_invalid_values
    total_empty_labels += split_empty_labels
    total_unreadable_images += split_unreadable_images

    # ========================================================
    # CLASS DISTRIBUTION
    # ========================================================

    print()
    print(f"{split.upper()} CLASS DISTRIBUTION")
    print("-" * 100)

    print(
        f"{'ID':<5}"
        f"{'Class':<15}"
        f"{'Objects':>12}"
        f"{'Images':>12}"
    )

    print("-" * 100)

    for class_id, class_name in enumerate(
        CLASS_NAMES
    ):

        print(
            f"{class_id:<5}"
            f"{class_name:<15}"
            f"{split_class_objects[class_id]:>12}"
            f"{split_class_images[class_id]:>12}"
        )

# ============================================================
# FINAL DATASET SUMMARY
# ============================================================

print()
print("=" * 100)
print("FINAL DATASET SUMMARY")
print("=" * 100)

print(
    f"Total images:              "
    f"{total_images}"
)

print(
    f"Total labels:              "
    f"{total_labels}"
)

print(
    f"Total objects:             "
    f"{total_objects}"
)

print(
    f"Expected images:           "
    f"{EXPECTED_TOTAL}"
)

print()

print(
    f"Invalid annotation lines:  "
    f"{total_invalid_lines}"
)

print(
    f"Invalid class IDs:         "
    f"{total_invalid_class_ids}"
)

print(
    f"Invalid YOLO values:       "
    f"{total_invalid_values}"
)

print(
    f"Empty label files:         "
    f"{total_empty_labels}"
)

print(
    f"Unreadable images:         "
    f"{total_unreadable_images}"
)

# ============================================================
# FINAL CLASS DISTRIBUTION
# ============================================================

print()
print("=" * 100)
print("FINAL CLASS DISTRIBUTION")
print("=" * 100)

print(
    f"{'ID':<5}"
    f"{'Class':<15}"
    f"{'Objects':>12}"
    f"{'Images':>12}"
)

print("-" * 100)

for class_id, class_name in enumerate(
    CLASS_NAMES
):

    print(
        f"{class_id:<5}"
        f"{class_name:<15}"
        f"{global_class_objects[class_id]:>12}"
        f"{global_class_images[class_id]:>12}"
    )

# ============================================================
# FINAL PASS / FAIL
# ============================================================

print()
print("=" * 100)
print("FINAL VALIDATION RESULT")
print("=" * 100)

errors = []

if total_images != EXPECTED_TOTAL:

    errors.append(
        f"Expected {EXPECTED_TOTAL} images, "
        f"found {total_images}"
    )

if total_labels != total_images:

    errors.append(
        f"Image/label mismatch: "
        f"{total_images} images vs "
        f"{total_labels} labels"
    )

if total_invalid_lines > 0:

    errors.append(
        f"{total_invalid_lines} invalid annotation lines"
    )

if total_invalid_class_ids > 0:

    errors.append(
        f"{total_invalid_class_ids} invalid class IDs"
    )

if total_invalid_values > 0:

    errors.append(
        f"{total_invalid_values} invalid YOLO values"
    )

if total_empty_labels > 0:

    errors.append(
        f"{total_empty_labels} empty label files"
    )

if total_unreadable_images > 0:

    errors.append(
        f"{total_unreadable_images} unreadable images"
    )

# ------------------------------------------------------------
# Check each split independently
# ------------------------------------------------------------

for split in SPLITS:

    images_dir = DATASET / split / "images"
    labels_dir = DATASET / split / "labels"

    if not images_dir.exists():
        continue

    if not labels_dir.exists():
        continue

    split_images = {
        p.stem
        for p in images_dir.iterdir()
        if p.is_file()
        and p.suffix.lower() in IMAGE_EXTENSIONS
    }

    split_labels = {
        p.stem
        for p in labels_dir.iterdir()
        if p.is_file()
        and p.suffix.lower() == ".txt"
    }

    if split_images != split_labels:

        errors.append(
            f"{split}: image/label filename mismatch"
        )

# ============================================================
# RESULT
# ============================================================

if errors:

    print()
    print("❌ VALIDATION FAILED")
    print()

    for error in errors:
        print(" -", error)

else:

    print()
    print("✅ VALIDATION PASSED")
    print()
    print("✓ 7,362 usable images")
    print("✓ 7,362 matching labels")
    print("✓ All image-label pairs match")
    print("✓ All images are readable")
    print("✓ All YOLO annotation lines are valid")
    print("✓ All class IDs are valid")
    print("✓ All normalized coordinates are valid")
    print("✓ No empty label files")
    print("✓ Train + valid + test = 7,362")
    print("✓ Dataset is ready for visual inspection")

print()
print("=" * 100)
print("VALIDATION COMPLETE")
print("=" * 100)
```

```python
import cv2
import matplotlib.pyplot as plt

# ============================================================
# DISPLAY FIRST 10 EXDARK TRAINING IMAGES WITH BOUNDING BOXES
# ============================================================

DATASET = Path(r"D:\ExDark_YOLO")

IMAGES_DIR = DATASET / "train" / "images"
LABELS_DIR = DATASET / "train" / "labels"

CLASS_NAMES = [
    "Bicycle",
    "Boat",
    "Bottle",
    "Bus",
    "Car",
    "Cat",
    "Chair",
    "Cup",
    "Dog",
    "Motorbike",
    "People",
    "Table"
]

IMAGE_EXTENSIONS = {
    ".jpg", ".jpeg", ".png", ".bmp", ".webp"
}

# ------------------------------------------------------------
# Get first 10 images alphabetically
# ------------------------------------------------------------

images = sorted([
    p for p in IMAGES_DIR.iterdir()
    if p.is_file()
    and p.suffix.lower() in IMAGE_EXTENSIONS
])[:10]

print(f"Displaying {len(images)} images:")
for image in images:
    print(" -", image.name)

# ------------------------------------------------------------
# Display
# ------------------------------------------------------------

fig, axes = plt.subplots(
    2, 5,
    figsize=(20, 9)
)

axes = axes.flatten()

for ax, image_path in zip(axes, images):

    # Read image
    image = cv2.imread(str(image_path))

    if image is None:
        ax.set_title("Could not read image")
        ax.axis("off")
        continue

    # Convert BGR -> RGB
    image = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2RGB
    )

    image_height, image_width = image.shape[:2]

    # Corresponding label
    label_path = LABELS_DIR / f"{image_path.stem}.txt"

    # Draw annotations
    if label_path.exists():

        with open(
            label_path,
            "r",
            encoding="utf-8"
        ) as f:

            lines = [
                line.strip()
                for line in f
                if line.strip()
            ]

        for line in lines:

            parts = line.split()

            if len(parts) != 5:
                continue

            class_id = int(parts[0])

            x_center = float(parts[1])
            y_center = float(parts[2])
            width = float(parts[3])
            height = float(parts[4])

            # Convert YOLO -> pixel coordinates
            x_center_px = x_center * image_width
            y_center_px = y_center * image_height

            box_width_px = width * image_width
            box_height_px = height * image_height

            x1 = int(
                x_center_px - box_width_px / 2
            )
            y1 = int(
                y_center_px - box_height_px / 2
            )
            x2 = int(
                x_center_px + box_width_px / 2
            )
            y2 = int(
                y_center_px + box_height_px / 2
            )

            # Keep box inside image
            x1 = max(0, min(x1, image_width - 1))
            y1 = max(0, min(y1, image_height - 1))
            x2 = max(0, min(x2, image_width - 1))
            y2 = max(0, min(y2, image_height - 1))

            # Draw rectangle
            cv2.rectangle(
                image,
                (x1, y1),
                (x2, y2),
                (255, 0, 0),
                2
            )

            # Class name
            class_name = (
                CLASS_NAMES[class_id]
                if 0 <= class_id < len(CLASS_NAMES)
                else f"Class {class_id}"
            )

            # Label
            cv2.putText(
                image,
                class_name,
                (x1, max(y1 - 5, 15)),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.5,
                (255, 0, 0),
                2,
                cv2.LINE_AA
            )

    # Display
    ax.imshow(image)
    ax.set_title(image_path.name, fontsize=9)
    ax.axis("off")

# Hide unused axes
for ax in axes[len(images):]:
    ax.axis("off")

plt.tight_layout()
plt.show()
```
